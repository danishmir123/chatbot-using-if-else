```python
# Simple chatbot using if-else

print("🤖 Chatbot: Hello! I’m ChatBot. Type 'bye' to exit.")

while True:
    user_input = input("me: ").lower()

    if user_input == "hello":
        print("🤖 Chatbot: Hi there!")
    elif user_input == "how are you":
        print("🤖 Chatbot: I'm just a bunch of code, but thanks for asking!")
    elif user_input == "what is your name":
        print("🤖 Chatbot: My name is ChatBot.")
    elif user_input == "bye":
        print("🤖 Chatbot: Goodbye! Have a nice day!")
        break
    elif "weather" in user_input:
        print("🤖 Chatbot: I can't check the weather now, but it's always sunny inside your heart!")
    else:
        print("🤖 Chatbot: Sorry, I don’t understand that.")
```
